### 2. Create Sales Column

```python
#Q4. Calculate sales for each transaction.
csv['sales']= csv['Quantity'] * csv['UnitPrice']
```

```python
csv
```

```python
#Q5. What is the total revenue?
csv['sales'].sum()
```

```python
#Q6. What is the average transaction value?
csv['sales'].mean()
```

### 3. Country Analysis

```python
#Q7. Which countries generate the most revenue?
csv.groupby('Country')['sales'].sum().sort_values(ascending=False)
```

```python
#Q8. Find the top 10 countries by revenue.
csv.groupby('Country')['sales'].sum().sort_values(ascending=False).head(10)
```

```python
#Q9. How many orders are there in each country?
csv.groupby('Country')['InvoiceNo'].nunique().sort_values(ascending=False)
```

```python
#Q10. Which are the top 10 products by revenue?
csv.groupby('Description')['sales'].sum().sort_values(ascending=False).head(10)
```

```python
# Q11. Which are the top 10 products by quantity sold?
csv.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)
```

### 5. Customer Analysis

```python
#Q12. Who are the top 10 customers by spending?
csv.groupby('CustomerID')['sales'].sum().sort_values(ascending=False).head(10)
```

```python
#Q13. How many orders did each customer place?
customer_orders = (csv.groupby('CustomerID')['InvoiceNo'].nunique().sort_values(ascending=False))
customer_orders
```

```python
#Q14. Find customers who placed more than 10 orders.
customer_orders[customer_orders>10]
```

```python
#Q15. What is the average spending per customer?
csv.groupby('CustomerID')['sales'].mean()
```

```python
customer_spending = csv.groupby('CustomerID')['sales'].sum()

customer_spending.mean()
```

### 6. Date Analysis

```python
#Q16. What is the total revenue for each month?
csv.groupby(csv['InvoiceDate'].dt.month)['sales'].sum()
```

```python
#right approch
monthly_sales = (
    csv.groupby(csv['InvoiceDate'].dt.to_period('M'))['sales'].sum()
)
monthly_sales
```

```python
#Q17. Which month had the highest revenue?
monthly_sales.idxmax()
```

```python
monthly_sales.max()
```

```python
monthly_sales.sort_values(ascending=False).head(1)
```
